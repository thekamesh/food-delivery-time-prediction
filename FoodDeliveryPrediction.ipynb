**Distance from coordinates — the Haversine formula**






The Haversine formula is a mathematical formula used to calculate the shortest distance between two points on the surface of a sphere, such as the Earth, using their latitude and longitude coordinates.

It calculates the great-circle distance between two geographical locations, assuming the Earth is approximately spherical.

*Haversine Formula

Let:

φ₁ = Latitude of Point 1
φ₂ = Latitude of Point 2
λ₁ = Longitude of Point 1
λ₂ = Longitude of Point 2
R = Radius of Earth ≈ 6371 km

First calculate:

$$ a = \sin^2\left(\frac{\phi_2-\phi_1}{2}\right) + \cos(\phi_1)\cos(\phi_2) \sin^2\left(\frac{\lambda_2-\lambda_1}{2}\right) $$

Then:

$$ c = 2\arctan2(\sqrt{a},\sqrt{1-a}) $$

Finally:

$$ \boxed{d = R \times c} $$

Where d is the distance between the two coordinates.**

```python
def calculate_series_distance(
    lat1_series, lon1_series, lat2_series, lon2_series, miles=False
):
    """Calculates the great-circle (straight-line) distance between two series of

    lat/lon points using the Haversine formula. Returns distance in km (or
    miles). Turns four raw coordinate columns into one meaningful 'distance'
    feature.
    """
    r = 3956 if miles else 6371
    lat1, lon1, lat2, lon2 = map(
        np.radians, [lat1_series, lon1_series, lat2_series, lon2_series]
    )
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    c = 2 * np.arcsin(np.sqrt(a))
    return c * r

```

**Encoding categorical variables
Categorical features come in two flavours, and each needs different treatment:

Ordinal (has a natural order): Road_traffic_density runs Low < Medium < High < Jam. We map it to 0, 1, 2, 3 so the model can learn a single "more traffic → more time" trend.
Nominal (no order): weather, city, vehicle type, order type. These get one-hot encoded into 0/1 columns. drop_first=True removes one redundant column per feature to avoid the dummy-variable trap (perfect multicollinearity).**

```python

# Traffic density has a real order (Low < Medium < High < Jam) -> ordinal encode instead of one-hot,
# so the model can learn a single, sensible increasing relationship
traffic_mapping={
    'Low':0,
    'Medium': 1,
    'High' : 2,
    'Jam' : 3
}

X['Road_traffic_density']=X['Road_traffic_density'].map(traffic_mapping)
```

**Model 1 — Linear Regression (baseline)
Simplest possible model: fits one straight-line relationship across all features. Fast and interpretable**

```python
from sklearn.linear_model import LinearRegression

```

```python
lr=LinearRegression()
lr.fit(X_train_scaled,y_train)
```

```python
lr.coef_
```

```python
from sklearn.metrics import r2_score, mean_squared_error

# Make predictions
y_train_pred = lr.predict(X_train_scaled)
y_test_pred = lr.predict(X_test_scaled)

# Calculate metrics for Train set
r2_train = r2_score(y_train, y_train_pred)
mse_train = mean_squared_error(y_train, y_train_pred)
rmse_train = np.sqrt(mse_train)

# Calculate metrics for Test set
r2_test = r2_score(y_test, y_test_pred)
mse_test = mean_squared_error(y_test, y_test_pred)
rmse_test = np.sqrt(mse_test)

print("=== Training Set Performance ===")
print(f"R2 Score: {r2_train:.4f}")
print(f"Mean Squared Error: {mse_train:.4f}")
print(f"Root Mean Squared Error: {rmse_train:.4f}")
print()
print("=== Testing Set Performance ===")
print(f"R2 Score: {r2_test:.4f}")
print(f"Mean Squared Error: {mse_test:.4f}")
print(f"Root Mean Squared Error: {rmse_test:.4f}")
```

**Model 2 —

 Decision Tree

```python

from sklearn.tree import DecisionTreeRegressor

dt=DecisionTreeRegressor(random_state=42)

dt.fit(X_train_scaled,y_train)
```

```python
y_pred_dt=dt.predict(X_test_scaled)
# Test performance
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
mse=mean_squared_error(y_test,dt.predict(X_test_scaled))
rmse=np.sqrt(mse)
r2=r2_score(y_test,dt.predict(X_test_scaled))
print("dt rmse",rmse)
print("dt r2",r2)
```

### **Model 3 — Random Forest Regressor**
An ensemble method that builds multiple decision trees on random subsets of the data and averages their predictions to reduce overfitting.

```python
from sklearn.ensemble import RandomForestRegressor

# Initialize Random Forest Regressor
rf = RandomForestRegressor(n_estimators=100, random_state=42)

# Fit the model
rf.fit(X_train_scaled, y_train)
```

```python
# Evaluate Random Forest on train and test data
y_train_pred_rf = rf.predict(X_train_scaled)
y_test_pred_rf = rf.predict(X_test_scaled)

mse_train_rf = mean_squared_error(y_train, y_train_pred_rf)
rmse_train_rf = np.sqrt(mse_train_rf)
r2_train_rf = r2_score(y_train, y_train_pred_rf)

mse_test_rf = mean_squared_error(y_test, y_test_pred_rf)
rmse_test_rf = np.sqrt(mse_test_rf)
r2_test_rf = r2_score(y_test, y_test_pred_rf)

print("=== Random Forest Train Performance ===")
print(f"RMSE: {rmse_train_rf:.4f}")
print(f"R2: {r2_train_rf:.4f}")
print()
print("=== Random Forest Test Performance ===")
print(f"RMSE: {rmse_test_rf:.4f}")
print(f"R2: {r2_test_rf:.4f}")
```

 **Model 4 — XGBoost
Trees trained sequentially — each new tree specifically corrects the errors of all trees before it (gradient boosting). Usually the strongest single model for structured/tabular data like ours.**

```python
from xgboost import XGBRegressor

xgb=XGBRegressor(n_estimators=100,random_state=42)
xgb.fit(X_train_scaled,y_train)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBRegressor

xgb_param_dist={
    'n_estimators':[300,500,800],
    'max_depth':[3,4,5,6,8],
    'learning_rate':[0.01,0.03,0.05,0.1],
    'subsample':[0.7,0.8,0.9,1.0],
    'colsample_bytree':[0.7,0.8,0.9,1.0],
    'reg_alpha':[0,0.1,0.5,1],
    'reg_lambda':[1,1.5,2]
}
xgb_search=RandomizedSearchCV(XGBRegressor(random_state=42),xgb_param_dist,n_iter=40,cv=3,scoring='r2',random_state=42,n_jobs=-1)
xgb_search.fit(X_train_scaled,y_train)
print("best params",xgb_search.best_params_)
xgb_best=xgb_search.best_estimator_
```

```python
r2_xgb_train=r2_score(y_train,xgb_best.predict(X_train_scaled))
rmse_xgb_train=np.sqrt(mean_squared_error(y_train,xgb_best.predict(X_train_scaled)))
print("xgb train rmse",rmse_xgb_train)
print("xgb train r2",r2_xgb_train)




```

## **Project Observations & Key Insights**

### **1. Data & Feature Insights**
* **Delivery Person Profile:**
  * Delivery partners with lower ratings (near `1.0`–`3.0`) correlate with significantly higher delivery times.
  * There is a steady, almost linear increase in delivery time as the delivery partner's age increases.
* **Environmental Factors:**
  * **Traffic:** Road traffic density is one of the strongest predictors. Under `Jam` conditions, delivery times jump dramatically to an average of `~31` minutes, compared to only `~21` minutes under `Low` traffic.
  * **Weather:** `Sunny` conditions offer the fastest deliveries (`~21.8` minutes), whereas `Fog` and `Cloudy` weather slow operations down to `~28.8` minutes on average.
  * **Festivals:** During festivals, delivery times surge dramatically to an average of `~45.5` minutes (compared to `~25.9` minutes normally), indicating high order volumes or staffing bottlenecks.
* **Geographical Distance:**
  * Raw coordinate data contained significant noise/outliers (absurd straight-line distances exceeding thousands of kilometers). Filtering for realistic intra-city trips (`< 100` km) was essential to stabilize our models.
